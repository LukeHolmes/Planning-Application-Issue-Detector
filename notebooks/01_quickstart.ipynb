# Quickstart — Planning Application Issue Detector

This notebook uses the `planning_detector` package (see repo README). Run from the repository root with the package installed: `pip install -e .`

```python
from pathlib import Path
import os

os.environ["PLANNING_DATA_PATH"] = "data/sample/planning_applications_sample.csv"

from planning_detector.config import PipelineConfig, default_data_path
from planning_detector.load import load_planning_data
from planning_detector.labels import build_labels
from planning_detector.train import train_pipeline

config = PipelineConfig.load()
path = default_data_path()
df = load_planning_data(path, config)
df.head()
```

```python
y = build_labels(df, config)
y.value_counts()
```

```python
_, metrics, meta = train_pipeline(path, config)
metrics
```
